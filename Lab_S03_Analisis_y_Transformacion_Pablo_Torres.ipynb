**1.1. Transformación de Variables Categóricas.**

Una variable categórica es aquella cuyo valores vienen fr un conjunto limitado de elementos.
Deben ser tratadas porque muchos de los algoritmos que usaré requieren que las variables sean numéricas

Pasos:
1. Borrar las variables categóricas que no aportan mucho valor al análisis

2. Label Encoding. Asigna a cada valor de la lista un número entero
   diferente.

3. One-hot-encoding. Crea nuevas columnas indicando la presencia de cada posible
   valor en el set de datos


```python

# aplicarlo al dataset real.
from numpy import array
from numpy import argmax
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import OneHotEncoder

# ejemolo con 3 categorias; frio templado y calor
data = ['frio', 'frio', 'templado', 'frio', 'calor', 'calor', 'templado','frio', 'templado', 'calor']
values = array(data)
print('Valores:',values)

# label encoding para convertir cada categoria de texto en un entero
label_encoder = LabelEncoder()
integer_encoded = label_encoder.fit_transform(values)
print('Label encoder integer:',integer_encoded)

# one hot encoding que a partir de los enteros crea una columna binaria por cada categoriap osible
onehot_encoder = OneHotEncoder(sparse_output=False)
integer_encoded = integer_encoded.reshape(len(integer_encoded), 1)
onehot_encoded = onehot_encoder.fit_transform(integer_encoded)
print('one-hot encoder:\n', onehot_encoded)

# inverse transform para volver del numero ya codificado al valor original categorico
inverted = label_encoder.inverse_transform([argmax(onehot_encoded[0, :])])
print(inverted) # MUESTRA LA VARIABLE QUE MÁS VECES SE TRANSFORMÓ.
```

3 valores distintos por lo que se sumaran 3 columas mostrando 1 o 0 para presencia o ausencia

```python

# aplicamos one hot encoding a la columna geography ya que tiene las 3 columnas que vimos antes
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [1])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
pd.DataFrame(X)

# mostramos el nuevo conjunto transformado con la variable geografia, observamos que ahora hay 3 columnas (1 por pais)
```
